```python
from uncertainty_manager import UncertaintyManager

print('Testing UncertaintyManager with system_failure_color_set.yaml...')
try:
    uncertainty_manager = UncertaintyManager('uncertainty_configs/system_failure_color_set.yaml')
    print('✅ UncertaintyManager initialized successfully!')
    
    # Print config info
    uncertainty_manager.print_config_info()
    
    # Test the color_set API
    print('\n🔍 Testing color_set API:')
    env_vars = uncertainty_manager.get_env_vars_for_api('color_set')
    print(f'   Environment variables: {env_vars}')
    
    uncertainty_types = uncertainty_manager.get_uncertainty_types_for_api('color_set')
    print(f'   Uncertainty types: {uncertainty_types}')
    
    has_uncertainties = uncertainty_manager.has_uncertainties_for_api('color_set')
    print(f'   Has uncertainties: {has_uncertainties}')
    
    # Test context manager
    print('\n🧪 Testing context manager:')
    with uncertainty_manager.apply_uncertainties_for_api('color_set') as uncertainty_info:
        print(f'   Applied uncertainties: {uncertainty_info}')
    
    print('\n✅ All tests passed!')

except Exception as e:
    print(f'❌ Error: {e}')
    import traceback
    traceback.print_exc()
```
